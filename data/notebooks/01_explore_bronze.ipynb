```python
import pandas as pd
from pathlib import Path

pd.set_option("display.max_columns", 200)
pd.set_option("display.width", 160)
```

```python
from pathlib import Path

PROJECT_ROOT = Path.cwd().parents[1] 
DATA_DIR = PROJECT_ROOT / "data"
```

```python

BRONZE_LAPS_BASE = PROJECT_ROOT / "data/bronze/laps/season=2024"
```

```python
rounds = sorted([p.name for p in BRONZE_LAPS_BASE.iterdir() if p.is_dir()])
rounds[:5], rounds[-5:], len(rounds)
```

```python
sample_rounds = [
    rounds[0],                 # first race
    rounds[len(rounds)//2],    # mid-season
    rounds[-1]                 # last race
]

sample_rounds
```

```python
r1_path = (
    BRONZE_LAPS_BASE
    / sample_rounds[0]
    / "session=R"
    / "data.parquet"
)

df_r1 = pd.read_parquet(r1_path)
df_r1.head(3)
```

```python
df_r1.dtypes
```

```python
rm_path = (
    BRONZE_LAPS_BASE
    / sample_rounds[1]
    / "session=R"
    / "data.parquet"
)

df_rm = pd.read_parquet(rm_path)
df_rm.dtypes
```

```python
rl_path = (
    BRONZE_LAPS_BASE
    / sample_rounds[2]
    / "session=R"
    / "data.parquet"
)

df_rl = pd.read_parquet(rl_path)
df_rl.dtypes
```

```python
set(df_r1.columns) == set(df_rm.columns) == set(df_rl.columns)
```
